# 01. Korea: gender equity at home and in the workplace, and fertility

This notebook tests the gender equity mismatch idea from McDonald (2000) and Goldin
(2024) within Korea. The idea is that fertility falls furthest where women gain close
to equal access to education and jobs in the public sphere while housework and
childcare at home stay traditional.

**Inputs** (all written by `00_pull_data.ipynb` to `data/raw/`): `korea_worldbank.csv`,
`korea_timeuse_kosis.csv`, `korea_parental_leave_moel.csv`.

**Outputs**: `data/processed/korea_combined.csv` (one row per year with every series),
figures `output/fig01_*.png`, and tables `output/table_korea_*.csv`.

| Sphere | Variable | Source |
|---|---|---|
| Outcome | Total fertility rate | World Bank |
| Public | Female labour force participation | World Bank / ILO |
| Public | Female tertiary enrolment | World Bank / UNESCO |
| Public | Female mean age at first marriage | World Bank |
| Private | Men's share of unpaid work | Statistics Korea Time Use Survey |
| Private | Fathers' share of parental leave takers | Ministry of Employment and Labor |

## 0. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from utils import (
    DATA_RAW, DATA_PROCESSED, share_pct, zscore, consecutive_changes,
    report_merge, save_figure, save_table,
)

pd.set_option("display.width", 120)
```

## 1. Load the raw files

The World Bank table already has one row per year. Men's share of unpaid work and
fathers' share of leave takers are computed here from the raw counts, as
`100 * men / (men + women)`. For 2010 and 2015 the leave share is the figure reported
directly in the press release, since counts by sex are incomplete for those years.

```python
public = pd.read_csv(DATA_RAW / "korea_worldbank.csv")
public["lfp_gap"] = public["male_lfp"] - public["female_lfp"]   # alternative public measure

tus = pd.read_csv(DATA_RAW / "korea_timeuse_kosis.csv")
tus["men_share_unpaid_pct"] = share_pct(tus["men_minutes"], tus["women_minutes"])

leave = pd.read_csv(DATA_RAW / "korea_parental_leave_moel.csv")
leave["father_share_leave_pct"] = share_pct(leave["fathers"], leave["mothers"]).fillna(
    leave["father_share_reported_pct"]
)

print("World Bank rows:", len(public), "| time use rows:", len(tus), "| leave rows:", len(leave))
display(tus[["year", "men_share_unpaid_pct"]].round(1))
display(leave[["year", "father_share_leave_pct"]].round(1))
```

Men's share of unpaid work rose from 10.6 percent in 1999 to 22.4 percent in 2024,
which is steady equalization, though women still do about 3.5 times as much housework.
Fathers went from about 3 percent of leave takers in 2010 to 31.6 percent in 2024. One
caution for the later sections is that uptake jumped after each expansion of benefits,
and those expansions were themselves responses to low fertility.

## 2. Combine everything into one table

Two left joins on `year` onto the World Bank table, then years without a fertility
number are dropped (that removes 2025).

```python
combined = public.merge(tus[["year", "men_share_unpaid_pct"]], on="year", how="left")
report_merge(public, tus, combined, "time use onto World Bank (left)", "year")

before = combined
combined = combined.merge(leave[["year", "father_share_leave_pct"]], on="year", how="left")
report_merge(before, leave, combined, "leave onto combined (left)", "year")

combined = combined.dropna(subset=["tfr"]).reset_index(drop=True)
print("Rows with a fertility value:", len(combined),
      "(", int(combined["year"].min()), "to", int(combined["year"].max()), ")")

save_table(combined, "korea_combined", folder=DATA_PROCESSED)
combined.tail(8)
```

## 3. The raw series

Four panels covering fertility, equity in the public sphere, equity in the private
sphere, and the timing of marriage.

Fertility fell in two episodes: a steep collapse to about 1.6 by the middle of the
1980s, and then a slower decline to 0.72. The equity series only begin between 1990
and 1999, after the first episode, so this data can speak only to the decline after
1999. Over that period both spheres improved while fertility kept falling, and the
age at marriage rose steadily. Marriage matters here because births outside marriage
are only 2 to 4 percent in Korea, so fertility moves closely with marriage.

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

ax = axes[0, 0]
ax.plot(combined["year"], combined["tfr"], color="black")
ax.axhline(2.1, color="grey", linestyle="--", linewidth=1)
ax.set_title("Total fertility rate (dashed = replacement)")
ax.set_ylabel("births per woman")

ax = axes[0, 1]
ax.plot(combined["year"], combined["female_lfp"], label="Female LFP, % (15+)")
ax.plot(combined["year"], combined["female_tertiary"], label="Female tertiary enrolment, %")
ax.set_title("Public-sphere equity")
ax.legend()

ax = axes[1, 0]
tus_points = combined.dropna(subset=["men_share_unpaid_pct"])
ax.plot(tus_points["year"], tus_points["men_share_unpaid_pct"], marker="o",
        label="Men's share of unpaid work, % (TUS)")
leave_points = combined.dropna(subset=["father_share_leave_pct"])
ax.plot(leave_points["year"], leave_points["father_share_leave_pct"], marker="s",
        label="Fathers' share of leave takers, %")
ax.axhline(50, color="grey", linestyle="--", linewidth=1)
ax.set_title("Private-sphere equity (dashed = parity)")
ax.set_ylim(0, 60)
ax.legend()

ax = axes[1, 1]
marriage_points = combined.dropna(subset=["female_marriage_age"])
ax.plot(marriage_points["year"], marriage_points["female_marriage_age"], marker="o", color="purple")
ax.set_title("Female mean age at first marriage")
ax.set_ylabel("years")

plt.tight_layout()
save_figure(fig, "fig01_korea_raw_series")
plt.show()
```

## 4. Everything on one scale

Each series is standardized by subtracting its own mean and dividing by its own
standard deviation. This compares timing and shape rather than magnitude, since a
small change and a large one both get stretched to a similar visual range.
Standardizing is used instead of rescaling to a range from 0 to 100, because that
approach pins each line to its two most extreme years.

The plot points to the core problem for any test within Korea. Every equity series
rises smoothly while fertility falls smoothly, so there is no distinctive matching of
turning points, just two opposite trends.

```python
columns_to_scale = ["tfr", "female_lfp", "female_tertiary", "men_share_unpaid_pct", "father_share_leave_pct"]
scaled = combined[["year"]].join(combined[columns_to_scale].apply(zscore))

fig, ax = plt.subplots(figsize=(12, 6))
for column in columns_to_scale:
    points = scaled.dropna(subset=[column])
    ax.plot(points["year"], points[column], marker=".", label=column)
ax.axhline(0, color="grey", linewidth=1)
ax.set_title("All series as z-scores (timing comparison only)")
ax.set_ylabel("standard deviations from own mean")
ax.legend()
save_figure(fig, "fig01_korea_zscores")
plt.show()
```

## 5. Correlations in levels

Correlating raw levels across years gives values near 0.9 in one direction or the
other for everything. That mostly reflects the fact that all the series trend in one
direction over the same decades, and two smooth trends will correlate strongly
whether or not they are related. The table is kept as a demonstration of the problem
rather than as evidence. The sign on the equity variables is negative, which read
literally would say that equalization lowers fertility, and that is a first hint that
this is only trend fitting.

```python
candidate_columns = ["female_lfp", "female_tertiary", "lfp_gap", "female_marriage_age",
                     "men_share_unpaid_pct", "father_share_leave_pct"]

level_table = pd.DataFrame({
    "variable": candidate_columns,
    "r_levels": [combined["tfr"].corr(combined[c]) for c in candidate_columns],
    "years_used": [combined[["tfr", c]].dropna().shape[0] for c in candidate_columns],
}).round(3)
level_table
```

## 6. Correlations in changes

The usual fix for trending series is to correlate changes from period to period
instead of levels, which removes the shared trend. The annual variables use changes
from one year to the next, keeping only differences that span exactly one year. The
Time Use Survey uses changes over five years matched to fertility changes over the
same five years.

```python
annual_columns = ["female_lfp", "female_tertiary", "lfp_gap", "father_share_leave_pct"]

rows = []
for column in annual_columns:
    changes = consecutive_changes(combined, ["tfr", column], step=1)
    rows.append({"variable": column,
                 "r_changes": changes["tfr"].corr(changes[column]),
                 "changes_used": len(changes)})

tus_changes = consecutive_changes(combined, ["tfr", "men_share_unpaid_pct"], step=5)
rows.append({"variable": "men_share_unpaid_pct (5-yearly)",
             "r_changes": tus_changes["tfr"].corr(tus_changes["men_share_unpaid_pct"]),
             "changes_used": len(tus_changes)})

difference_table = pd.DataFrame(rows).round(3)
difference_table
```

The level correlations near 0.9 collapse to roughly zero once we look at changes,
falling to 0.07 for labour force participation, 0.01 for tertiary enrolment and 0.01
for parental leave. This suggests the level correlations were mostly artifacts of the
trend, and that short term movements in the equity variables have no detectable
relationship with short term movements in fertility. The Time Use row of 0.53 in the
negative direction rests on only five changes, so it should not be leaned on either
way.

```python
# Table 3 in the paper: levels and changes side by side.
table3 = level_table.merge(
    difference_table.assign(variable=lambda d: d["variable"].str.replace(" (5-yearly)", "", regex=False)),
    on="variable", how="left",
)
save_table(table3, "table_korea_correlations")
table3
```

## 7. Lag check

If equity affects fertility with a delay, changes should correlate at some positive
lag. This scans lags of zero to four years, but only for series where every lag keeps
at least 15 observations, since scanning many lags on a handful of changes will tend
to find something by chance. Only female labour force participation and tertiary
enrolment qualify, as the leave and time use series are too short for this.

The noise bar is `2 / sqrt(n)`, roughly the size a correlation of pure noise reaches
one time in twenty.

```python
def lag_scan(var_column, data, max_lag):
    """Correlate the yearly change in TFR at year t with the yearly change in
    `var_column` at year t - lag, for lag = 0..max_lag."""
    d_tfr = consecutive_changes(data, ["tfr"], step=1)["tfr"]
    d_var = consecutive_changes(data, [var_column], step=1)[var_column]

    rows = []
    for lag in range(0, max_lag + 1):
        shifted = d_var.copy()
        shifted.index = shifted.index + lag          # change at year y now lines up with TFR change at y + lag
        pair = pd.concat([d_tfr.rename("d_tfr"), shifted.rename("d_var")], axis=1, join="inner")
        n = len(pair)
        if n < 3:
            continue
        r = pair["d_tfr"].corr(pair["d_var"])
        bar = 2.0 / np.sqrt(n)
        rows.append({"lag": lag, "correlation": round(r, 3), "n": n,
                     "noise_bar": round(bar, 2), "clears_bar": abs(r) > bar})
    return pd.DataFrame(rows)


print("Female LFP (lag in years):")
display(lag_scan("female_lfp", combined, max_lag=4))
print("Female tertiary enrolment (lag in years):")
display(lag_scan("female_tertiary", combined, max_lag=4))
```

Nothing here changes the story. Labour force participation shows no relationship at
any lag. Tertiary enrolment drifts upward with the lag and reaches 0.47 at four
years, but it peaks at the edge of the scanned range and carries a sign that runs
backwards to the theory, since faster growth in enrolment appears to predict growth
in fertility. That looks more like leftover curvature in the trend than a real
delayed effect, so it is treated as a null result and set aside.

## 8. The mismatch score

The hypothesis is about the gap between the two spheres, so each sphere gets a small
composite. The public composite averages the standardized labour force participation
and tertiary enrolment. The private composite averages the standardized time use
share and the leave share, interpolated to annual values. The mismatch is the public
composite minus the private one.

The gap in labour force participation is left out because it carries nearly the same
signal as female participation, and the age at marriage is left out because it is an
outcome rather than a measure of equity. Interior gaps in the time use and leave
series are filled in by linear interpolation.

```python
work = combined[["year", "tfr", "female_lfp", "female_tertiary",
                 "men_share_unpaid_pct", "father_share_leave_pct"]].sort_values("year").reset_index(drop=True)

# Fill interior gaps only (no extrapolation past the first or last survey).
work["tus_interp"] = work["men_share_unpaid_pct"].interpolate(method="linear", limit_area="inside")
work["leave_interp"] = work["father_share_leave_pct"].interpolate(method="linear", limit_area="inside")

# z-score each component over the years it exists.
z = work[["female_lfp", "female_tertiary", "tus_interp", "leave_interp"]].apply(zscore)
z.columns = ["z_lfp", "z_tertiary", "z_tus", "z_leave"]
work = work.join(z)

# Public composite needs both components; private composite averages whatever exists.
work["public_composite"] = work[["z_lfp", "z_tertiary"]].mean(axis=1, skipna=False)
work["private_composite"] = work[["z_tus", "z_leave"]].mean(axis=1)   # NaN only if both missing
work["private_composition"] = np.select(
    [work["z_tus"].isna() & work["z_leave"].isna(), work["z_leave"].isna(), work["z_tus"].isna()],
    ["none", "tus_only", "leave_only"],
    default="both",
)
work["mismatch_score"] = work["public_composite"] - work["private_composite"]

print("Private composite composition by period:")
print(work.groupby("private_composition")["year"].agg(["min", "max", "count"]))

both = work.dropna(subset=["mismatch_score", "tfr"])
print()
print("Years with a full mismatch score:", int(both["year"].min()), "to", int(both["year"].max()),
      "(", len(both), "years )")
r_mismatch = both["tfr"].corr(both["mismatch_score"])
print("Correlation between composite mismatch and TFR (levels, descriptive only):", round(r_mismatch, 3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

ax = axes[0]
for column, label, color in [("public_composite", "Public equity composite", None),
                             ("private_composite", "Private equity composite", None)]:
    points = work.dropna(subset=[column])
    ax.plot(points["year"], points[column], label=label, color=color)
ax.plot(work["year"], zscore(work["tfr"]), color="black", label="TFR (z)")
ax.axhline(0, color="grey", linewidth=1)
ax.set_title("Composites over time")
ax.legend()

ax = axes[1]
ax.scatter(both["mismatch_score"], both["tfr"])
for _, row in both.iterrows():
    ax.annotate(str(int(row["year"])), (row["mismatch_score"], row["tfr"]), fontsize=8)
ax.set_xlabel("Mismatch (public composite - private composite)")
ax.set_ylabel("Total fertility rate")
ax.set_title(f"Composite mismatch vs fertility, 2000-2024 (r = {r_mismatch:.2f}, descriptive)")

plt.tight_layout()
save_figure(fig, "fig01_korea_composite_mismatch")
plt.show()
```

The result comes out positive at about 0.4, meaning that a falling mismatch coincided
with falling fertility, which is the opposite of what the theory predicts. The time
plot shows why. After 2015 the private composite outpaced the public one, driven
almost entirely by the surge in parental leave, which was itself a policy response to
low fertility. Two things follow from that. First, a standardized mismatch measured
within one country captures the pace along each sphere's own trajectory rather than
the absolute gap the theory is about. Second, variables that track policy uptake are
shaped by fertility itself and do not belong in an index meant to explain it. The
absolute gap can only be tested across countries.

## 9. Takeaways

- The mismatch configuration is real. Korea combines close to equal education and
  employment with a housework split of roughly 3.5 to 1, so the precondition of the
  hypothesis holds.
- Within Korea, though, the hypothesis cannot be confirmed. Levels correlate
  spuriously, changes show nothing, lags show nothing robust, and the composite
  mismatch comes out with the wrong sign for reasons that look artifactual.
- The private sphere equalized throughout the collapse, moving from 10.6 to 22.4
  percent, which works against the simplest version of the story in which more
  equality quickly brings more births. It remains consistent with threshold or
  generational versions, though those cannot be tested here.
- Fertility runs through marriage in Korea, since births outside marriage are only 2
  to 4 percent and the age at marriage keeps rising. That suggests the mechanism, if
  it is real, works on the forming of marriages rather than on births within
  marriage.
- The total fertility rate is a period measure that is pushed down by postponement,
  and completed fertility by cohort is also falling, so the decline is real but worth
  a caveat.
- Overall, one country's time series cannot separate these explanations. The version
  of the test based on the absolute gap needs variation across countries, which is
  the subject of `02_oecd_mismatch.ipynb`.
