# 00. Pull all raw data

This notebook pulls every dataset used in the project and saves it to `data/raw/`.
Nothing is analysed here. The three analysis notebooks (`01` to `03`) read the
files this notebook writes, so the whole project can be re-run from scratch by
running this notebook once and then the other three in order.

| File written | Contents | Source | Used by |
|---|---|---|---|
| `korea_worldbank.csv` | Korea, one row per year 1980–2025: TFR, female/male LFP, female tertiary enrolment, female mean age at first marriage | World Bank API | 01 |
| `korea_timeuse_kosis.csv` | Korea, minutes per day of unpaid work by sex, 1999–2024 every five years | Statistics Korea Time Use Survey via KOSIS (hand entered) | 01 |
| `korea_parental_leave_moel.csv` | Korea, parental leave benefit recipients by sex, 2010–2024 | Ministry of Employment and Labor press release (hand entered) | 01 |
| `oecd_worldbank_annual.csv` | 38 OECD members, long format, full annual series: TFR, F/M LFP ratio, tertiary gender parity index, GDP per capita PPP | World Bank API | 02 |
| `oecd_time_use.csv` | 30 OECD members, minutes per day of unpaid work by sex, one survey per country | OECD Time Use Database (SDMX API) | 02 |
| `oecd_births_outside_marriage.csv` | OECD members, latest share of births outside marriage and its year | OECD Family Database table SF2.4 | 03 |
| `SF_2_4_Share_births_outside_marriage.xlsx` | The original SF2.4 workbook, unmodified | OECD Family Database | 03 |

No API key is needed for any of these.

## 0. Setup

```python
import pandas as pd

from utils import (
    DATA_RAW, OECD_MEMBERS, WB_INDICATORS_KOREA, WB_INDICATORS_OECD, KOREA_CODE,
    pull_world_bank_panel, pull_oecd_time_use, pull_oecd_births_outside_marriage,
    save_table,
)

pd.set_option("display.width", 120)
print("Raw data will be written to:", DATA_RAW)
```

## 1. Korea, World Bank annual series

- `SP.DYN.TFRT.IN` total fertility rate
- `SL.TLF.CACT.FE.ZS` and `SL.TLF.CACT.MA.ZS` female and male labour force participation, age 15+
- `SE.TER.ENRR.FE` female gross tertiary enrolment, which can pass 100 percent because it is a gross ratio
- `SP.DYN.SMAM.FE` female mean age at marriage, reported every five years and ending in 2015

Korea's fertility decline is mostly a post-1980 story, so the series is cut at 1980.
The table is saved wide, one row per year, with missing values left blank.

```python
START_YEAR = 1980

korea_long = pull_world_bank_panel([KOREA_CODE], WB_INDICATORS_KOREA)
korea_long = korea_long[korea_long["year"] >= START_YEAR]

korea = korea_long.pivot(index="year", columns="indicator", values="value").reset_index()
korea.columns.name = None
korea = korea[["year"] + list(WB_INDICATORS_KOREA)]   # fixed column order

print("Non-missing years per series:")
print(korea.drop(columns="year").notna().sum())
save_table(korea, "korea_worldbank", folder=DATA_RAW)
korea.tail()
```

## 2. Korea, Time Use Survey (hand entered)

Daily minutes of housework and family care for married people living with their
spouse, averaged across all days. The figures from 1999 to 2014 come from a KOSIS
table on time spent by marital status for people aged 20 and above, and the figures
from 2019 to 2024 come from Statistics Korea press releases. The two sources agree
exactly in 2014, so they chain together cleanly. These numbers were typed in by hand
because KOSIS has no open API for this table; the `source` column records where each
row came from.

```python
korea_timeuse = pd.DataFrame({
    "year":          [1999, 2004, 2009, 2014, 2019, 2024],
    "men_minutes":   [35,   39,   44,   50,   60,   65],
    "women_minutes": [295,  277,  267,  259,  244,  225],
    "source": ["KOSIS table"] * 4 + ["Statistics Korea press release"] * 2,
})
save_table(korea_timeuse, "korea_timeuse_kosis", folder=DATA_RAW)
korea_timeuse
```

## 3. Korea, parental leave recipients by sex (hand entered)

Recipients of parental leave benefits for the first time, by sex, from employment
insurance data published by the Ministry of Employment and Labor in February 2025.
Counts by sex are only available from 2017 (fathers from 2015). For 2010 and 2015 the
press release reports the fathers' share directly, so that share is stored in its own
column and the counts are left blank.

```python
korea_leave = pd.DataFrame({
    "year":    [2010, 2015, 2017,  2018,  2019,  2020,  2021,  2022,  2023,  2024],
    "fathers": [None, 4872, 12042, 17662, 22297, 27421, 29041, 37884, 35336, 41829],
    "mothers": [None, None, 78080, 81537, 82868, 84617, 81514, 93200, 90672, 90706],
    "father_share_reported_pct": [2.7, 5.6, None, None, None, None, None, None, None, None],
})
save_table(korea_leave, "korea_parental_leave_moel", folder=DATA_RAW)
korea_leave
```

## 4. OECD, World Bank annual series for all 38 members

The same API, four indicators, every year available. Saved long (one row per
country, indicator and year) so notebook 02 can both take the latest value per
country and draw the time plots.

```python
oecd_panel = pull_world_bank_panel(list(OECD_MEMBERS), WB_INDICATORS_OECD)
oecd_panel["country"] = oecd_panel["code"].map(OECD_MEMBERS)
oecd_panel = oecd_panel[["code", "country", "indicator", "year", "value"]]

print("Rows:", len(oecd_panel), "| countries:", oecd_panel["code"].nunique(),
      "| indicators:", oecd_panel["indicator"].nunique())
print("Latest year available per indicator:")
print(oecd_panel.groupby("indicator")["year"].max())
save_table(oecd_panel, "oecd_worldbank_annual", folder=DATA_RAW)
```

## 5. OECD Time Use Database

Minutes per day of unpaid work (housework, care and shopping) for men and women aged
15 to 64, one survey per country, through the OECD SDMX API. The dataflow does not
return the survey year, so the age of each snapshot is unknown; the OECD
documentation puts them roughly between 2009 and 2024.

```python
oecd_timeuse = pull_oecd_time_use()
missing = sorted(set(OECD_MEMBERS) - set(oecd_timeuse["code"]))
print("OECD members with time-use data:", len(oecd_timeuse))
print("Members without time-use data:", ", ".join(OECD_MEMBERS[c] for c in missing))
save_table(oecd_timeuse, "oecd_time_use", folder=DATA_RAW)
oecd_timeuse.head()
```

## 6. OECD Family Database, share of births outside marriage (SF2.4)

The OECD publishes this as an Excel workbook with one row per country and one
column per year from 1960 to 2024. The latest non-missing value per country is kept,
and the original workbook is saved alongside it. The OECD notes that for Australia,
Japan, Korea and New Zealand the figure counts births where the parents are not
registered as married to each other, while for everyone else it counts births to
mothers whose legal status is other than married.

```python
nonmarital = pull_oecd_births_outside_marriage(
    save_raw_to=DATA_RAW / "SF_2_4_Share_births_outside_marriage.xlsx"
)
print("OECD members with a value:", len(nonmarital))
print("Year of latest value:")
print(nonmarital["nonmarital_year"].value_counts().sort_index())
save_table(nonmarital, "oecd_births_outside_marriage", folder=DATA_RAW)
```
