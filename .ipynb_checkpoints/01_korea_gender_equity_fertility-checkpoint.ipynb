# Korea: gender equity at home and in the workplace, and fertility

This notebook tests the gender equity mismatch idea from McDonald 2001 and Goldin
2024 within Korea. The idea is that fertility falls furthest where women gain close
to equal access to education and jobs in the public sphere while housework and
childcare at home stay traditional.

| Sphere | Variable | Source |
|---|---|---|
| Outcome | Total fertility rate | World Bank |
| Public | Female labour force participation | World Bank / ILO |
| Public | Female tertiary enrolment | World Bank / UNESCO |
| Public | Female mean age at first marriage | World Bank |
| Private | Men's share of unpaid work | Statistics Korea Time Use Survey |
| Private | Fathers' share of parental leave takers | Ministry of Employment and Labor |
| Private | Husband's share of housework | KLIPS panel |

## 0. Setup

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# Years we care about. Korea's fertility decline is mostly a post-1980 story.
START_YEAR = 1980
END_YEAR = 2025

pd.set_option("display.width", 120)
```

## 1. Helper to pull one indicator from the World Bank API

A free JSON API that needs no key. It returns a year and a value for one indicator
in one country.


```python
def get_world_bank_series(country_code, indicator_code):
    """Return a DataFrame with columns 'year' and 'value' for one indicator."""

    url = "https://api.worldbank.org/v2/country/" + country_code + "/indicator/" + indicator_code
    parameters = {"format": "json", "per_page": 200}

    response = requests.get(url, params=parameters)
    response.raise_for_status()
    payload = response.json()

    # The payload is a list of two things: [metadata, records]
    records = payload[1]

    years = []
    values = []
    for record in records:
        if record["value"] is None:
            continue
        years.append(int(record["date"]))
        values.append(float(record["value"]))

    result = pd.DataFrame({"year": years, "value": values})
    result = result.sort_values("year")
    result = result[result["year"] >= START_YEAR]
    result = result[result["year"] <= END_YEAR]
    result = result.reset_index(drop=True)
    return result
```

## 2. Outcome and public sphere variables

- `SP.DYN.TFRT.IN` for the total fertility rate
- `SL.TLF.CACT.FE.ZS` and `SL.TLF.CACT.MA.ZS` for female and male labour force
  participation from age 15 up
- `SE.TER.ENRR.FE` for female gross tertiary enrolment, which can pass 100 percent
  because it is a gross ratio
- `SP.DYN.SMAM.FE` for female mean age at marriage, reported every five years and
  ending in 2015


```python
tfr = get_world_bank_series("KOR", "SP.DYN.TFRT.IN")
female_lfp = get_world_bank_series("KOR", "SL.TLF.CACT.FE.ZS")
male_lfp = get_world_bank_series("KOR", "SL.TLF.CACT.MA.ZS")
female_tertiary = get_world_bank_series("KOR", "SE.TER.ENRR.FE")
female_marriage_age = get_world_bank_series("KOR", "SP.DYN.SMAM.FE")

print("TFR rows:", len(tfr))
print("Female LFP rows:", len(female_lfp))
print("Female tertiary rows:", len(female_tertiary))
print("Marriage age rows:", len(female_marriage_age))

tfr.tail()
```

One table keyed by year. The gap in labour force participation between men and
women is also computed as an alternative measure of the public sphere.


```python
public = pd.DataFrame({"year": range(START_YEAR, END_YEAR + 1)})

public = public.merge(tfr.rename(columns={"value": "tfr"}), on="year", how="left")
public = public.merge(female_lfp.rename(columns={"value": "female_lfp"}), on="year", how="left")
public = public.merge(male_lfp.rename(columns={"value": "male_lfp"}), on="year", how="left")
public = public.merge(female_tertiary.rename(columns={"value": "female_tertiary"}), on="year", how="left")
public = public.merge(female_marriage_age.rename(columns={"value": "female_marriage_age"}), on="year", how="left")

public["lfp_gap"] = public["male_lfp"] - public["female_lfp"]

public.tail(10)
```

## 3. Private sphere variables

### 3a. Time Use Survey

Daily minutes of housework and family care for married people living with their
spouse, averaged across all days. The figures from 1999 to 2014 come from a KOSIS
table on time spent by marital status for people aged 20 and above, and the figures
from 2019 to 2024 come from Statistics Korea press releases. The two sources agree
exactly in 2014, so they chain together cleanly.

Men's share rose from 10.6 percent in 1999 to 22.4 percent in 2024, which is steady
equalization, though women still do about 3.5 times as much housework.


```python
# Minutes per day on unpaid work, married people living with spouse, all-days average.
# None = not yet found (KOSIS time-series table).
tus_years = [1999, 2004, 2009, 2014, 2019, 2024]
tus_men_minutes = [35, 39, 44, 50, 60, 65]
tus_women_minutes = [295, 277, 267, 259, 244, 225]

tus = pd.DataFrame({
    "year": tus_years,
    "men_minutes": tus_men_minutes,
    "women_minutes": tus_women_minutes,
})

# Men's share of total unpaid work, as a percentage.
men_share_list = []
for i in range(len(tus)):
    men = tus.loc[i, "men_minutes"]
    women = tus.loc[i, "women_minutes"]
    if pd.isna(men) or pd.isna(women):
        men_share_list.append(None)
    else:
        share = 100.0 * men / (men + women)
        men_share_list.append(share)

tus["men_share_unpaid_pct"] = men_share_list
tus
```

### 3b. Fathers' share of parental leave takers

Recipients of leave benefits for the first time by sex, drawn from employment
insurance data published by the Ministry of Employment and Labor in February 2025.
Only shares are available for 2010 and 2015. Fathers went from about 3 percent of
takers in 2010 to 31.6 percent in 2024. One caution for the later sections is that
uptake jumped after each expansion of benefits, and those expansions were themselves
responses to low fertility.


```python
leave_years   = [2010, 2015, 2017,  2018,  2019,  2020,  2021,  2022,  2023,  2024]
leave_fathers = [None, 4872, 12042, 17662, 22297, 27421, 29041, 37884, 35336, 41829]
leave_mothers = [None, None, 78080, 81537, 82868, 84617, 81514, 93200, 90672, 90706]

# Shares reported directly for years where counts by sex are incomplete.
known_share = {2010: 2.7, 2015: 5.6}

leave = pd.DataFrame({
    "year": leave_years,
    "fathers": leave_fathers,
    "mothers": leave_mothers,
})

father_share_list = []
for i in range(len(leave)):
    year = leave.loc[i, "year"]
    fathers = leave.loc[i, "fathers"]
    mothers = leave.loc[i, "mothers"]
    if pd.notna(fathers) and pd.notna(mothers):
        share = 100.0 * fathers / (fathers + mothers)
        father_share_list.append(share)
    elif year in known_share:
        father_share_list.append(known_share[year])
    else:
        father_share_list.append(None)

leave["father_share_leave_pct"] = father_share_list
leave
```

### 3c. KLIPS annual housework hours

The Korean Labor and Income Panel Study has run every year since 1998 and would give
about 25 annual observations of the husband's share of housework, but it requires
registration. The file is used here if it is present, and skipped otherwise.


```python
# FILL IN: the column names in your exported KLIPS CSV
KLIPS_FILE = "klips_housework.csv"
COL_YEAR = "year"
COL_HOUSEHOLD = "hhid"
COL_SEX = "sex"            # assumed coded 1 = male, 2 = female (KLIPS convention)
COL_HOUSEWORK = "housework_hours_per_week"

MALE_CODE = 1
FEMALE_CODE = 2

klips_available = False
try:
    klips_raw = pd.read_csv(KLIPS_FILE)
    klips_available = True
    print("Loaded KLIPS file with", len(klips_raw), "rows")
except FileNotFoundError:
    print("KLIPS file not found. Skipping section 3c for now.")
```

```python
klips = None

if klips_available:
    # Keep only the columns we need and drop rows with missing housework hours.
    klips_small = klips_raw[[COL_YEAR, COL_HOUSEHOLD, COL_SEX, COL_HOUSEWORK]].copy()
    klips_small = klips_small.dropna()

    # Split into men and women.
    men_rows = klips_small[klips_small[COL_SEX] == MALE_CODE]
    women_rows = klips_small[klips_small[COL_SEX] == FEMALE_CODE]

    # Average hours per year for each sex.
    men_by_year = men_rows.groupby(COL_YEAR)[COL_HOUSEWORK].mean()
    women_by_year = women_rows.groupby(COL_YEAR)[COL_HOUSEWORK].mean()

    years_list = []
    men_hours_list = []
    women_hours_list = []
    share_list = []

    for year in sorted(men_by_year.index):
        if year not in women_by_year.index:
            continue
        men_hours = men_by_year[year]
        women_hours = women_by_year[year]
        share = 100.0 * men_hours / (men_hours + women_hours)

        years_list.append(int(year))
        men_hours_list.append(men_hours)
        women_hours_list.append(women_hours)
        share_list.append(share)

    klips = pd.DataFrame({
        "year": years_list,
        "klips_men_hours": men_hours_list,
        "klips_women_hours": women_hours_list,
        "klips_men_share_pct": share_list,
    })

    display(klips)
```

## 4. Combine everything into one table

```python
combined = public.copy()

combined = combined.merge(tus[["year", "men_share_unpaid_pct"]], on="year", how="left")
combined = combined.merge(leave[["year", "father_share_leave_pct"]], on="year", how="left")

if klips is not None:
    combined = combined.merge(klips[["year", "klips_men_share_pct"]], on="year", how="left")

# Keep only years where we at least have a fertility number.
combined = combined.dropna(subset=["tfr"])
combined = combined.reset_index(drop=True)

combined.tail(10)
```

## 5. The raw series

Four panels covering fertility, equity in the public sphere, equity in the private
sphere, and the timing of marriage.

Fertility fell in two episodes: a steep collapse to about 1.6 by the middle of the
1980s, and then a slower decline to 0.72. The equity series only begin between 1990
and 1999, after the first episode, so this data can speak only to the decline after
1999. Over that period both spheres improved while fertility kept falling, and the
age at marriage rose steadily. Marriage matters here because births outside marriage
are only 2 to 4 percent in Korea, so fertility moves closely with marriage.


```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Panel 1: fertility
ax = axes[0, 0]
ax.plot(combined["year"], combined["tfr"], color="black")
ax.axhline(2.1, color="grey", linestyle="--", linewidth=1)
ax.set_title("Total fertility rate (dashed = replacement)")
ax.set_ylabel("births per woman")

# Panel 2: public sphere
ax = axes[0, 1]
ax.plot(combined["year"], combined["female_lfp"], label="Female LFP, % (15+)")
ax.plot(combined["year"], combined["female_tertiary"], label="Female tertiary enrolment, %")
ax.set_title("Public-sphere equity")
ax.legend()

# Panel 3: private sphere
ax = axes[1, 0]
tus_points = combined.dropna(subset=["men_share_unpaid_pct"])
ax.plot(tus_points["year"], tus_points["men_share_unpaid_pct"], marker="o", label="Men's share of unpaid work, % (TUS)")

leave_points = combined.dropna(subset=["father_share_leave_pct"])
ax.plot(leave_points["year"], leave_points["father_share_leave_pct"], marker="s", label="Fathers' share of leave takers, %")

if "klips_men_share_pct" in combined.columns:
    klips_points = combined.dropna(subset=["klips_men_share_pct"])
    ax.plot(klips_points["year"], klips_points["klips_men_share_pct"], marker=".", label="Men's share of housework hours, % (KLIPS)")

ax.axhline(50, color="grey", linestyle="--", linewidth=1)
ax.set_title("Private-sphere equity (dashed = parity)")
ax.set_ylim(0, 60)
ax.legend()

# Panel 4: marriage timing
ax = axes[1, 1]
marriage_points = combined.dropna(subset=["female_marriage_age"])
ax.plot(marriage_points["year"], marriage_points["female_marriage_age"], marker="o", color="purple")
ax.set_title("Female mean age at first marriage")
ax.set_ylabel("years")

plt.tight_layout()
plt.show()
```

## 6. Everything on one scale

Each series is standardized by subtracting its own mean and dividing by its own
standard deviation. This compares timing and shape rather than magnitude, since a
small change and a large one both get stretched to a similar visual range.
Standardizing is used instead of rescaling to a range from 0 to 100, because that
approach pins each line to its two most extreme years.

The plot points to the core problem for any test within Korea. Every equity series
rises smoothly while fertility falls smoothly, so there is no distinctive matching of
turning points, just two opposite trends.


```python
def zscore(series):
    """Standardize a series: (value - mean) / standard deviation. Missing values stay missing."""
    average = series.mean()
    spread = series.std()
    if spread == 0:
        return series * 0
    standardized = (series - average) / spread
    return standardized


scaled = pd.DataFrame({"year": combined["year"]})

columns_to_scale = ["tfr", "female_lfp", "female_tertiary", "men_share_unpaid_pct", "father_share_leave_pct"]
if "klips_men_share_pct" in combined.columns:
    columns_to_scale.append("klips_men_share_pct")

for column in columns_to_scale:
    scaled[column] = zscore(combined[column])

plt.figure(figsize=(12, 6))
for column in columns_to_scale:
    points = scaled.dropna(subset=[column])
    if len(points) == 0:
        continue
    plt.plot(points["year"], points[column], marker=".", label=column)

plt.axhline(0, color="grey", linewidth=1)
plt.title("All series as z-scores (timing comparison only)")
plt.ylabel("standard deviations from own mean")
plt.legend()
plt.show()
```

## 7. Correlations in levels

Correlating raw levels across years gives values near 0.9 in one direction or the
other for everything. That mostly reflects the fact that all the series trend in one
direction over the same decades, and two smooth trends will correlate strongly
whether or not they are related. The table is kept as a demonstration of the problem
rather than as evidence. The sign on the equity variables is negative, which read
literally would say that equalization lowers fertility, and that is a first hint that
this is only trend fitting.


```python
candidate_columns = ["female_lfp", "female_tertiary", "lfp_gap", "female_marriage_age",
                     "men_share_unpaid_pct", "father_share_leave_pct"]
if "klips_men_share_pct" in combined.columns:
    candidate_columns.append("klips_men_share_pct")

names = []
correlations = []
counts = []

for column in candidate_columns:
    pair = combined[["tfr", column]].dropna()
    number_of_years = len(pair)
    if number_of_years < 3:
        continue
    r = pair["tfr"].corr(pair[column])
    names.append(column)
    correlations.append(round(r, 3))
    counts.append(number_of_years)

correlation_table = pd.DataFrame({
    "variable": names,
    "correlation_with_tfr": correlations,
    "years_used": counts,
})
correlation_table
```

## 7b. Correlations in changes

The usual fix for trending series is to correlate changes from period to period
instead of levels, which removes the shared trend. The annual variables use changes
from one year to the next, and the Time Use Survey uses changes over five years
matched to fertility changes over the same five years.


```python
# Year-over-year differences for the annual variables.
annual_columns = ["female_lfp", "female_tertiary", "lfp_gap", "father_share_leave_pct"]
if "klips_men_share_pct" in combined.columns:
    annual_columns.append("klips_men_share_pct")

diff_names = []
diff_correlations = []
diff_counts = []

for column in annual_columns:
    pair = combined[["year", "tfr", column]].dropna()
    pair = pair.sort_values("year").reset_index(drop=True)

    # Keep only consecutive years so a difference always spans exactly one year.
    tfr_changes = []
    var_changes = []
    for i in range(1, len(pair)):
        if pair.loc[i, "year"] - pair.loc[i - 1, "year"] != 1:
            continue
        tfr_changes.append(pair.loc[i, "tfr"] - pair.loc[i - 1, "tfr"])
        var_changes.append(pair.loc[i, column] - pair.loc[i - 1, column])

    if len(tfr_changes) < 3:
        continue

    changes = pd.DataFrame({"d_tfr": tfr_changes, "d_var": var_changes})
    r = changes["d_tfr"].corr(changes["d_var"])
    diff_names.append(column)
    diff_correlations.append(round(r, 3))
    diff_counts.append(len(changes))

# 5-year differences for the TUS series.
tus_pair = combined[["year", "tfr", "men_share_unpaid_pct"]].dropna()
tus_pair = tus_pair.sort_values("year").reset_index(drop=True)

tfr_changes = []
var_changes = []
for i in range(1, len(tus_pair)):
    tfr_changes.append(tus_pair.loc[i, "tfr"] - tus_pair.loc[i - 1, "tfr"])
    var_changes.append(tus_pair.loc[i, "men_share_unpaid_pct"] - tus_pair.loc[i - 1, "men_share_unpaid_pct"])

if len(tfr_changes) >= 3:
    changes = pd.DataFrame({"d_tfr": tfr_changes, "d_var": var_changes})
    r = changes["d_tfr"].corr(changes["d_var"])
    diff_names.append("men_share_unpaid_pct (5-yearly)")
    diff_correlations.append(round(r, 3))
    diff_counts.append(len(changes))

difference_table = pd.DataFrame({
    "variable": diff_names,
    "correlation_of_changes": diff_correlations,
    "changes_used": diff_counts,
})
difference_table
```

The level correlations near 0.9 collapse to roughly zero once we look at changes,
falling to 0.07 for labour force participation, 0.01 for tertiary enrolment and 0.01
for parental leave. This suggests the level correlations were mostly artifacts of the
trend, and that short term movements in the equity variables have no detectable
relationship with short term movements in fertility. The Time Use row of 0.53 in the
negative direction rests on only five changes, so it should not be leaned on either
way.


## 7c. Lag check

If equity affects fertility with a delay, changes should correlate at some positive
lag. This scans lags of zero to four years, but only for series where every lag keeps
at least 15 observations, since scanning many lags on a handful of changes will tend
to find something by chance. Only female labour force participation and tertiary
enrolment qualify, as the leave and time use series are too short for this.


```python
def differences_for(column, data, step):
    """Return years and changes, keeping only differences that span exactly `step` years."""
    pair = data[["year", column]].dropna()
    pair = pair.sort_values("year").reset_index(drop=True)
    change_years = []
    changes = []
    for i in range(1, len(pair)):
        if pair.loc[i, "year"] - pair.loc[i - 1, "year"] != step:
            continue
        change_years.append(pair.loc[i, "year"])
        changes.append(pair.loc[i, column] - pair.loc[i - 1, column])
    return change_years, changes


def lag_scan(var_column, data, max_lag):
    """Correlate the yearly change in TFR at time t with the yearly change in the
    variable at t - lag, for lag = 0..max_lag."""
    tfr_years, tfr_changes = differences_for("tfr", data, 1)
    var_years, var_changes = differences_for(var_column, data, 1)

    results = []
    for lag in range(0, max_lag + 1):
        matched_tfr = []
        matched_var = []
        for i in range(len(var_years)):
            target_year = var_years[i] + lag
            if target_year in tfr_years:
                j = tfr_years.index(target_year)
                matched_tfr.append(tfr_changes[j])
                matched_var.append(var_changes[i])
        n = len(matched_tfr)
        if n < 3:
            continue
        frame = pd.DataFrame({"d_tfr": matched_tfr, "d_var": matched_var})
        r = frame["d_tfr"].corr(frame["d_var"])
        bar = 2.0 / (n ** 0.5)
        results.append({"lag": lag, "correlation": round(r, 3),
                        "n": n, "noise_bar": round(bar, 2),
                        "clears_bar": abs(r) > bar})
    return pd.DataFrame(results)


# Only the two series long enough for a lag scan (>= 15 observations at every lag).
# The leave and TUS series are too short: scanning lags there would be curve-fitting.
print("Female LFP (lag in years):")
display(lag_scan("female_lfp", combined, max_lag=4))

print("Female tertiary enrolment (lag in years):")
display(lag_scan("female_tertiary", combined, max_lag=4))
```

Nothing here changes the story. Labour force participation shows no relationship at
any lag. Tertiary enrolment drifts upward with the lag and reaches 0.47 at four
years, but it peaks at the edge of the scanned range and carries a sign that runs
backwards to the theory, since faster growth in enrolment appears to predict growth
in fertility. That looks more like leftover curvature in the trend than a real
delayed effect, so it is treated as a null result and set aside.


## 8. The mismatch score

The hypothesis is about the gap between the two spheres, so each sphere gets a small
composite. The public composite averages the standardized labour force participation
and tertiary enrolment. The private composite averages the standardized time use
share and the leave share, interpolated to annual values. The mismatch is the public
composite minus the private one.

The gap in labour force participation is left out because it carries nearly the same
signal as female participation, and the age at marriage is left out because it is an
outcome rather than a measure of equity. Interior gaps in the time use and leave
series are filled in by linear interpolation.


```python
# Interpolate the TUS men's share to annual frequency.
work = combined[["year", "tfr", "female_lfp", "female_tertiary",
                 "men_share_unpaid_pct", "father_share_leave_pct"]].copy()
work = work.sort_values("year").reset_index(drop=True)

work["tus_interp"] = work["men_share_unpaid_pct"].interpolate(method="linear",
                                                              limit_area="inside")

# The leave series also has interior gaps (2011-2014, 2016); interpolate those too,
# so the private composite doesn't flip composition year to year.
work["leave_interp"] = work["father_share_leave_pct"].interpolate(method="linear",
                                                                  limit_area="inside")

# z-score each component over the years it exists.
work["z_lfp"] = zscore(work["female_lfp"])
work["z_tertiary"] = zscore(work["female_tertiary"])
work["z_tus"] = zscore(work["tus_interp"])
work["z_leave"] = zscore(work["leave_interp"])

# Public composite: needs both components.
public_list = []
for i in range(len(work)):
    z1 = work.loc[i, "z_lfp"]
    z2 = work.loc[i, "z_tertiary"]
    if pd.isna(z1) or pd.isna(z2):
        public_list.append(None)
    else:
        public_list.append((z1 + z2) / 2.0)
work["public_composite"] = public_list

# Private composite: average of what exists (TUS-only before 2010).
private_list = []
composition_note = []
for i in range(len(work)):
    z_tus = work.loc[i, "z_tus"]
    z_leave = work.loc[i, "z_leave"]
    if pd.isna(z_tus) and pd.isna(z_leave):
        private_list.append(None)
        composition_note.append("none")
    elif pd.isna(z_leave):
        private_list.append(z_tus)
        composition_note.append("tus_only")
    elif pd.isna(z_tus):
        private_list.append(z_leave)
        composition_note.append("leave_only")
    else:
        private_list.append((z_tus + z_leave) / 2.0)
        composition_note.append("both")
work["private_composite"] = private_list
work["private_composition"] = composition_note

print("Private composite composition by period:")
print(work.groupby("private_composition")["year"].agg(["min", "max", "count"]))

work["mismatch_score"] = work["public_composite"] - work["private_composite"]

both = work.dropna(subset=["mismatch_score", "tfr"])
print()
print("Years with a full mismatch score:", int(both["year"].min()), "to", int(both["year"].max()),
      "(", len(both), "years )")
r = both["tfr"].corr(both["mismatch_score"])
print("Correlation between composite mismatch and TFR (levels, descriptive only):", round(r, 3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Left: the two composites and TFR (z-scored for comparability) over time.
ax = axes[0]
plot_data = work.dropna(subset=["public_composite"])
ax.plot(plot_data["year"], plot_data["public_composite"], label="Public equity composite")
plot_data = work.dropna(subset=["private_composite"])
ax.plot(plot_data["year"], plot_data["private_composite"], label="Private equity composite")
plot_data = work.dropna(subset=["tfr"])
ax.plot(plot_data["year"], zscore(work["tfr"]), color="black", label="TFR (z)")
ax.axhline(0, color="grey", linewidth=1)
ax.set_title("Composites over time")
ax.legend()

# Right: mismatch vs TFR scatter.
ax = axes[1]
ax.scatter(both["mismatch_score"], both["tfr"])
for i in both.index:
    ax.annotate(str(int(both.loc[i, "year"])),
                (both.loc[i, "mismatch_score"], both.loc[i, "tfr"]), fontsize=8)
ax.set_xlabel("Mismatch (public composite - private composite)")
ax.set_ylabel("Total fertility rate")
ax.set_title("Composite mismatch vs fertility (descriptive)")

plt.tight_layout()
plt.show()
```

The result comes out positive at about 0.4, meaning that a falling mismatch coincided
with falling fertility, which is the opposite of what the theory predicts. The time
plot shows why. After 2015 the private composite outpaced the public one, driven
almost entirely by the surge in parental leave, which was itself a policy response to
low fertility. Two things follow from that. First, a standardized mismatch measured
within one country captures the pace along each sphere's own trajectory rather than
the absolute gap the theory is about. Second, variables that track policy uptake are
shaped by fertility itself and do not belong in an index meant to explain it. The
absolute gap can only be tested across countries.


## 9. Takeaways

- The mismatch configuration is real. Korea combines close to equal education and
  employment with a housework split of roughly 3.5 to 1, so the precondition of the
  hypothesis holds.
- Within Korea, though, the hypothesis cannot be confirmed. Levels correlate
  spuriously, changes show nothing, lags show nothing robust, and the composite
  mismatch comes out with the wrong sign for reasons that look artifactual.
- The private sphere equalized throughout the collapse, moving from 10.6 to 22.4
  percent, which works against the simplest version of the story in which more
  equality quickly brings more births. It remains consistent with threshold or
  generational versions, though those cannot be tested here.
- Fertility runs through marriage in Korea, since births outside marriage are only 2
  to 4 percent and the age at marriage keeps rising. That suggests the mechanism, if
  it is real, works on the forming of marriages rather than on births within
  marriage.
- The total fertility rate is a period measure that is pushed down by postponement,
  and completed fertility by cohort is also falling, so the decline is real but worth
  a caveat.
- Overall, one country's time series cannot separate these explanations. The version
  of the test based on the absolute gap needs variation across countries, which is
  the subject of part two.
